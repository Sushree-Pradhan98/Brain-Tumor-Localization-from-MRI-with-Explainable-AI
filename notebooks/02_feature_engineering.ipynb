## Feature Engineering

In this project, feature engineering focuses on preparing MRI images and segmentation masks to improve model performance.

###  Normalization
Pixel values were scaled to the range **[0, 1]** to stabilize training and improve convergence.

###  Resizing
All images and masks were resized to a uniform shape (e.g., **256 x 256**) to ensure compatibility with deep learning models.

###  Mask Processing
Segmentation masks were converted into binary format, where tumor regions are labeled as **1** and background as **0**.

###  Data Augmentation
Data augmentation is applied during training to improve model generalization. Since MRI images in the BraTS dataset exhibit variability in tumor size, shape, and location, augmentation helps the model learn robust features.

Transformations such as horizontal flipping, small rotations, and brightness/contrast adjustments are applied. These augmentations are carefully chosen to preserve anatomical correctness.

Augmentation is performed dynamically during training on preprocessed (.npy) images, ensuring efficient memory usage and increased variability across epochs.

---

These steps enhance the model's ability to learn meaningful patterns while handling variability in tumor size and appearance.
